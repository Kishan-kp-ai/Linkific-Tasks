```python
import pandas as pd

df = pd.read_csv("student_data.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nStatistical Summary:")
df.describe()
```

```python
df[df.isnull().any(axis=1)]
```

```python
# Fill missing numerical values with median
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())

# Fill missing categorical value with mode
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Check missing values again
print(df.isnull().sum())
```

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("\nStudents by Gender:")
print(df["Gender"].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(df["Marks"], bins=5, kde=True)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="City", y="Marks")

plt.title("Average Marks by City")
plt.xlabel("City")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="Gender", y="Marks")

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Age", y="Marks")

plt.title("Age vs Student Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.show()
```

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

print("\nAverage Marks by City:")
print(df.groupby("City")["Marks"].mean())

print("\nAverage Marks by Gender:")
print(df.groupby("Gender")["Marks"].mean())
```
