```python
import numpy as np

# Student marks matha,science,englih
marks = np.array([
    [85, 90, 78],
    [72, 68, 80],
    [95, 88, 92],
    [60, 75, 70],
    [88, 82, 85]
])

# Student-wise calculations
total_marks = np.sum(marks, axis=1)
average_marks = np.mean(marks, axis=1)

# Subject-wise calculations
subject_average = np.mean(marks, axis=0)

# Overall analysis
highest_mark = np.max(marks)
lowest_mark = np.min(marks)
top_student = np.argmax(average_marks)

print("STUDENT MARKS ANALYSIS")
print("----------------------")

print("Total Marks:", total_marks)
print("Average Marks:", average_marks)

print("\nSubject Averages:")
print("Math:", subject_average[0])
print("Science:", subject_average[1])
print("English:", subject_average[2])

print("\nHighest Mark:", highest_mark)
print("Lowest Mark:", lowest_mark)

print("\nTop Student:", top_student + 1)
print("Top Student Average:", average_marks[top_student])
```
