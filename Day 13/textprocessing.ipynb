```python
# Import libraries
import pandas as pd
import nltk
import re
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Create a small text dataset

texts = [
    "Machine learning is a powerful technology.",
    "Python is widely used for machine learning.",
    "Machine learning helps computers learn from data.",
    "Data science uses Python for data analysis.",
    "Artificial intelligence is changing the technology world.",
    "Python is easy to learn and useful for data science.",
    "Machine learning models can make predictions.",
    "Data is important for building machine learning models.",
    "Artificial intelligence and machine learning are growing rapidly.",
    "Learning Python can help students build AI projects."
]

df = pd.DataFrame({"Text": texts})

print(df)
```

```python
# Convert text to lowercase

df["Lowercase_Text"] = df["Text"].str.lower()

print(df[["Text", "Lowercase_Text"]])
```

```python
import nltk
nltk.download('punkt')

# Tokenization
df["Tokens"] = df["Lowercase_Text"].apply(word_tokenize)

print(df[["Lowercase_Text", "Tokens"]])
```

```python
# Get English stopwords

stop_words = set(stopwords.words("english"))

# Remove stopwords

df["Filtered_Tokens"] = df["Tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

print(df[["Tokens", "Filtered_Tokens"]])
```

```python
# Join filtered tokens back into text

df["Processed_Text"] = df["Filtered_Tokens"].apply(
    lambda tokens: " ".join(tokens)
)

print(df[["Text", "Processed_Text"]])
```

```python
# Create TF-IDF Vectorizer

vectorizer = TfidfVectorizer()

# Convert processed text into numerical values

tfidf_matrix = vectorizer.fit_transform(df["Processed_Text"])
```

```python
# Display feature names

feature_names = vectorizer.get_feature_names_out()

print("Features:")
print(feature_names)
```

```python
# Create TF-IDF DataFrame

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names
)

print("\nTF-IDF Matrix:")
print(tfidf_df.round(2))
```

```python
print("\nTotal text documents:", len(df))
print("Total unique words:", len(feature_names))
```
