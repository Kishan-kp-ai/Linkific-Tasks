```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_student_data.csv")

df.head()
```

```python
df.describe()
```

```python
print("Gender Distribution:")
print(df["Gender"].value_counts())

print("\nCity Distribution:")
print(df["city"].value_counts())
```

```python
#Average marks and attendance
print("Average Marks:", df["marks"].mean())
print("Average Attendance:", df["attendance"].mean())
```

```python
#Highest and lowest performers
print("Highest Marks:")
print(df.loc[df["marks"].idxmax()])

print("\nLowest Marks:")
print(df.loc[df["marks"].idxmin()])
```

```python
#Check whether attendance and marks are related
correlation = df["marks"].corr(df["attendance"])

print("Correlation between Marks and Attendance:", correlation)
```

```python
#Compare performance by gender
df.groupby("Gender")[["marks", "attendance"]].mean()
```

```python
#Marks Distribution
plt.figure(figsize=(8, 5))

plt.hist(df["marks"], bins=10, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
#Average Marks by Gender — Bar Chart
gender_marks = df.groupby("Gender")["marks"].mean()

plt.figure(figsize=(7, 5))

plt.bar(gender_marks.index, gender_marks.values)

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.show()
```

```python
#Marks vs Attendance — Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(df["attendance"], df["marks"])

plt.title("Marks vs Attendance")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

plt.show()
```

```python
#Average Marks by City — Bar Chart
city_marks = df.groupby("city")["marks"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))

plt.bar(city_marks.index, city_marks.values)

plt.title("Average Marks by City")
plt.xlabel("City")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.show()
```

```python
#important  insight
print("Overall Summary")
print("----------------")
print("Average Marks:", round(df["marks"].mean(), 2))
print("Average Attendance:", round(df["attendance"].mean(), 2))
print("Highest Marks:", df["marks"].max())
print("Lowest Marks:", df["marks"].min())

print("\nAverage Marks by Gender")
print("-----------------------")
print(df.groupby("Gender")["marks"].mean().round(2))

print("\nAverage Marks by City")
print("---------------------")
print(df.groupby("city")["marks"].mean().round(2))

print("\nMarks-Attendance Correlation")
print("---------------------------")
print(round(df["marks"].corr(df["attendance"]), 2))
```
