```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END


# -------------------------
# State Management
# -------------------------

class State(TypedDict):
    count: int
    status: str


# -------------------------
# Node 1: Process
# -------------------------

def process(state: State):
    print("Processing...")

    state["count"] += 1

    # Simulate success after 3 attempts
    if state["count"] >= 3:
        state["status"] = "success"
    else:
        state["status"] = "retry"

    return state


# -------------------------
# Conditional Routing
# -------------------------

def check_status(state: State):

    if state["status"] == "success":
        return "success"

    return "retry"


# -------------------------
# Create Graph
# -------------------------

graph = StateGraph(State)

graph.add_node("process", process)

graph.add_edge(START, "process")

graph.add_conditional_edges(
    "process",
    check_status,
    {
        "success": END,
        "retry": "process"
    }
)

app = graph.compile()


# -------------------------
# Run Workflow
# -------------------------

result = app.invoke({
    "count": 0,
    "status": ""
})

print("\nFinal State:")
print(result)
```
