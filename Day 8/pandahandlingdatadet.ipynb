```python
import pandas as pd

df = pd.read_csv("student_data_65_students_cleaning_project.csv")

df
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df["marks"] = df["marks"].fillna(df["marks"].median())
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
```

```python
df.isnull().sum()
```

```python
#dublicate records
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
#feature engineering
df.columns
```

```python
df = df.rename(columns={
    "Student Name": "name",
    "age": "age",
    "marks": "marks",
    "city": "city",
    "Attendance": "attendance"
})
```

```python
df.columns
```

```python
#handel data type
df.dtypes
```

```python

df["attendance"] = pd.to_numeric(df["attendance"], errors="coerce")
df["attendance"] = df["attendance"].fillna(df["attendance"].median())
df.dtypes
```

```python
df.to_csv("cleaned_student_data.csv", index=False)
```

```python
print("Shape:", df.shape)
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
