```python
from transformers import pipeline
```

```python
# Text Generation Model

generator = pipeline(
    "text-generation",
    model="gpt2"
)

prompt = "Artificial Intelligence is changing the way"

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

print("Input:")
print(prompt)

print("\nGenerated Text:")
print(result[0]["generated_text"])
```

```python
# Sentiment Analysis Model

sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

text = "I really enjoyed learning about artificial intelligence today."

result = sentiment_model(text)

print("Input:")
print(text)

print("\nSentiment:")
print(result)
```

```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6"
)

text = """
Artificial intelligence has developed significantly over the past few decades.
Machine learning allows computers to learn patterns from data without being
explicitly programmed for every task. Deep learning uses neural networks with
multiple layers to solve complex problems. More recently, generative AI models
have become capable of creating text, images, code, audio and other types of
content. Large language models are now being used for question answering,
content generation, summarization and programming assistance.
"""

result = summarizer(
    text,
    max_length=80,
    min_length=30,
    do_sample=False
)

print("Summary:")
print(result[0]["summary_text"])
```
