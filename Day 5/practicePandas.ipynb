```python
import pandas as pd

df = pd.read_csv("student_data.csv")
print("Dataset loaded successfully!")
```

```python
df
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Last 5 rows:")
display(df.tail())
```

```python
rows, columns = df.shape
print(df.shape)
print("Number of rows:", rows)
print("Number of columns:", columns)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Students with marks greater than 80:")
display(df[df["Marks"] > 80])

print("Students older than 21:")
display(df[df["Age"] > 21])

print("Female students:")
display(df[df["Gender"] == "Female"])

print("Students from Hubli:")
display(df[df["City"] == "Hubli"])

print("Female students with marks greater than 80:")
display(df[(df["Gender"] == "Female") & (df["Marks"] > 80)])
```

```python
print("Summary Statistics:")
display(df.describe())
```
